```python
import csv
import os

file = "students.csv"

if not os.path.exists(file):
    f = open(file, "w", newline="")
    w = csv.writer(f)
    w.writerow(["Roll No", "Name", "Age", "Course"])
    f.close()


def add_student():
    roll = input("Enter Roll No: ")
    name = input("Enter Name: ")
    age = input("Enter Age: ")
    course = input("Enter Course: ")

    f = open(file, "a", newline="")
    w = csv.writer(f)
    w.writerow([roll, name, age, course])
    f.close()

    print("Student Added Successfully!")

def display_student():
    f = open(file, "r")
    print(f.read())
    f.close()


def search_student():
    roll = input("Enter Roll No: ")

    f = open(file, "r")
    data = f.read()
    f.close()

    if roll in data:
        print("Student Record Found!")
        print(data[data.find(roll):])
    else:
        print("Student Not Found!")

print("\n===== STUDENT DATABASE =====")
print("1. Add Student")
print("2. Display Student")
print("3. Search Student")

choice = input("Enter Your Choice: ")

if choice == "1":
    add_student()

elif choice == "2":
    display_student()

elif choice == "3":
    search_student()

else:
    print("Invalid Choice!")
```
